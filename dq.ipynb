## multiple run

```python
e_values = [0.2]  
lambda_values = [0.1]
alpha_values = [0.1]
delta_values = [0.99]
random_uniform_options = [True]
tolerance = [1e-3]
M = [100]

# Create a list to store explorations
explorations = []

# Generate combinations of parameters
parameter_combinations = itertools.product(
    e_values, lambda_values, alpha_values, delta_values, random_uniform_options, tolerance, M 
)

# Use tqdm to create a progress bar
for params in tqdm(parameter_combinations, total=len(e_values) * len(lambda_values) * len(alpha_values) * len(delta_values) * len(random_uniform_options* len(tolerance)*len(M))):
    e, lambda_val, alpha, delta, random_uniform, t, m = params
    
    # Call your function with the current combination of parameters
    result = algorithm_new(e, lambda_val, alpha, delta, random_uniform, t, m, verbose = False)
    
    # Store the combination and result
    explorations.append({
        
        'e': e,
        'lambda_val': lambda_val,
        'alpha': alpha,
        'delta': delta,
        'random_uniform': random_uniform,
        'tolerance': t,
        'M':m,
        'eqC': result[0],
        'eqD': result[1],
        'tot': (np.sum(result[0]) + np.sum(result[1]))
    })
 
```

```python
expl = pd.DataFrame(explorations)
```
