```python
expl = expl[expl['delta'] == 0.95]
expl = expl[expl['e'] == 0.2]
expl = expl[expl['tolerance'] == 0.00010]
expl = expl[expl['M'] == 100]
```

```python
pairs = list(itertools.combinations(['lambda_val', 'alpha'], 2))
for pair in pairs:
    pivot_table = expl.pivot_table(values='eqC', index=pair[0], columns=pair[1], aggfunc='mean')

    plt.figure(figsize=(10, 6))
    sns.heatmap(pivot_table, annot=True, cmap='coolwarm')
    plt.title(f'Heatmap of eqC by parameters: {pair[0]} and {pair[1]}')
    plt.xlabel(pair[1])
    plt.ylabel(pair[0])
    plt.show()
```
